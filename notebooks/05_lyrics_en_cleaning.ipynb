```python
# lyrics for analysis: prefer translated English, otherwise fall back to original lyrics
import duckdb as dd
from pathlib import Path

con = dd.connect()

PROJECT_ROOT = Path.cwd().parent  # go up from notebooks -> project root

source_path = PROJECT_ROOT / "data/processed/lyrics/2026/03/05/lyrics_lang_trans.csv"
output_path = PROJECT_ROOT / "data/processed/lyrics/2026/03/05/lyrics_in_en.csv"

df_lyrics_in_en = con.execute(f"""
    select
        rank,
        artist,
        title,
        region,
        spotify_uri,
        coalesce(lyrics_en, lyrics) as lyrics_in_en,
        language as original_lang
    from read_csv_auto('{source_path}', delim=',')
""").df()

# save as csv
df_lyrics_in_en.to_csv(f"{output_path}", index=False)

display(df_lyrics_in_en)
print(f"Wrote {len(df_lyrics_in_en):,} rows to {output_path}")
```
