```python
# lyrics for analysis, translated to English when possible
import duckdb as dd
from pathlib import Path

con = dd.connect()

file_path = "/Users/wednesday/Documents/GitHub/lyrics_analysis/data/processed/lyrics/2026/03/05/lyrics_lang_trans.csv"
output_path = Path(file_path).with_name("lyrics_lang_en.csv")

df_lyrics_en = con.execute(f""" 
    select spotify_uri, language, 
    case
        when language is null or lower(language) = 'en' then lyrics
        else lyrics_en
    end as lyrics_en
    from read_csv_auto('{file_path}')
""").df()

# save as csv
df_lyrics_en.to_csv(f"{output_path}", index=False)

display(df_lyrics_en)
print(f"Wrote {len(df_lyrics_en):,} rows to {output_path}")
```
