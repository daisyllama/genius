```python
import pandas as pd

file1 = r"D:\Users\Documents\GitHub\lyrics_analysis\data\processed\co_global_tw_usa\lyrics_lang.csv"
file2 = r"D:\Users\Documents\GitHub\lyrics_analysis\data\processed\sg_es_ar_jp\02_lyrics_lang.csv"

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

print(f"Length of df1: {len(df1)}")
display(df1.head())
print(f"\n Length of df2: {len(df2)}")
display(df2.head())
```

```python
df1["length"] = df1['lyrics'].str.len().astype('Int64')
df1.rename(columns={"language": "original_lang"}, inplace=True)
df1.drop(columns=["rank","region"], inplace=True)

display(df1.head())
```

```python
df_appended = pd.concat([df1, df2[df1.columns]], ignore_index=True)

print(f"\n Length of appended DataFrame: {len(df_appended)}")
display(df_appended.head())
```

```python
new_file = r"D:\Users\Documents\GitHub\lyrics_analysis\data\processed\new_02_lyrics_lang.csv"

df_appended.to_csv(new_file, index=False)
```
