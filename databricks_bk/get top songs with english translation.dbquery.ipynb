```sql
with view as (
select 
    rank, artist, title, region, spotify_uri,
    case 
        when language = 'en' then lyrics
        when lyrics_en is null then null
        else lyrics_en 
    end as lyrics_en
from songs_db.default.lyrics_lang_trans
)
select 
    row_number() over (partition by region order by rank asc) as rn,
    rank, artist, title, region, lyrics_en
from view
```
